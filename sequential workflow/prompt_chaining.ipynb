```python
from langgraph.graph import StateGraph,START,END
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from typing import TypedDict
load_dotenv()
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    
)
```

```python
class llm_state(TypedDict):
    title:str
    outline:str
    blog:str
```

```python
def create_outline(state: llm_state)->llm_state:
    title=state['title']
    prompt=f"write the outline of the given topic {title} for a blog"
    answer=llm.invoke(prompt).content
    state['outline']=answer
    return state
```

```python
def create_blog(state: llm_state)->llm_state:
    title=state['title']
    outline=state["outline"]

    prompt=f"write the blog on the given topic {title} with the help of given {outline}"
    answer=llm.invoke(prompt).content
    state['outline']=answer
    return state
```

```python
graph=StateGraph(llm_state)
graph.add_node("create_outline",create_outline)
graph.add_node("create_blog",create_blog)


graph.add_edge(START,"create_outline")
graph.add_edge("create_outline","create_blog")
graph.add_edge("create_blog",END)

workflow=graph.compile()
```

```python
topic={"title":"raise of Ai in pakistan"}
workflow.invoke(topic)
```
